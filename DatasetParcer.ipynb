```python
import numpy as np
import pandas as pd
import os
```

```python
curr_path = os.getcwd()
datasets_folder = os.path.join(curr_path, 'Datasets')
dataset_file = os.path.join(datasets_folder, 'MIPT_hackathon_dataset.csv')
os.path.exists(dataset_file)
```

```python
'''Таблица на самом деле вполне нормальная, просто необходимо уточнить разделитель (sep) и квалификатор текста (quotechar).
Например, состав заказа прописан в виде нескольких переносов строк с пунктами, двоеточиями, запятыми внутри и т. д.,
однако этот столбец целиком обособлен от всех остальных двойными кавычками. Бонусная информация: судя по всему, последним пунктом заказа
всегда значится Доставка (Артикул: 1 ...). ts - timestamp (полагаю, в секундах с 01.01.1970). Наверное, время с точностью до секунды
стоит привести в формат '%Y-%m-%d %H:%M:%S'
'''
dataset = pd.read_csv(dataset_file, sep = ',', quotechar='"', parse_dates = True, date_format = '%Y-%m-%d')
dataset.head(5)
```

```python
def epSec2Datetime(df):
    for col in df.columns.tolist():
        df[col] = pd.to_datetime(df[col], unit = 's')
    return df
def datetime2EpSec(df):
    for col in df.columns.tolist():
        df[col] = (df[col] - pd.Timestamp("1970-01-01")) // pd.Timedelta('1s')
    return df
```

```python
from pprint import pprint
columns = dataset.columns.tolist()
secCols = [i for i in columns if ('_ts' in i) or ('_at' in i) or ('Дата' in i)]
#pprint(secCols)
```

```python
dataset[secCols] = epSec2Datetime(dataset[secCols])
```
